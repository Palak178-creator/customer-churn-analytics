```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

sns.set_style("whitegrid")

print("Libraries Loaded Successfully")
```

```python
df = pd.read_csv("../data/processed/cleaned_telco_churn.csv")

df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.columns
```

```python
df.info()
```

```python
df["Churn"].value_counts()
```

```python
churn_percent = df["Churn"].value_counts(normalize=True) * 100

print(churn_percent)
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="Churn"
)

plt.title("Customer Churn Distribution")
plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x="gender",
    hue="Churn"
)

plt.title("Gender vs Churn")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Contract",
    hue="Churn"
)

plt.title("Contract Type vs Churn")
plt.xticks(rotation=10)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="tenure",
    bins=30
)

plt.title("Customer Tenure Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Churn",
    y="tenure"
)

plt.title("Tenure vs Churn")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Churn",
    y="MonthlyCharges"
)

plt.title("Monthly Charges vs Churn")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Churn",
    y="TotalCharges"
)

plt.title("Total Charges vs Churn")
plt.show()
```

```python
df["Churn_Binary"] = df["Churn"].map({
    "No": 0,
    "Yes": 1
})

df.head()
```

```python
numeric_cols = [
    "SeniorCitizen",
    "tenure",
    "MonthlyCharges",
    "TotalCharges",
    "Churn_Binary"
]

corr = df[numeric_cols].corr()

corr
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

# Key Findings

1. Month-to-month customers exhibit the highest churn rate.

2. Customers with lower tenure are more likely to churn.

3. Higher monthly charges are associated with increased churn.

4. Long-term customers are significantly more loyal.

5. Contract type appears to be one of the strongest churn indicators.
